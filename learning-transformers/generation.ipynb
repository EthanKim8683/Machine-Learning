```python
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL_ID = "HuggingFaceTB/SmolLM-135M"

model = AutoModelForCausalLM.from_pretrained(MODEL_ID)
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

text = "What is the capital of France?"
inputs = tokenizer(text, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=20, use_cache=False)
tokenizer.decode(outputs, skip_special_tokens=True)[0]
```

```python
import torch

def generate(input_ids, attention_mask = None, max_new_tokens = None):
    if attention_mask is None:
        attention_mask = torch.ones_like(input_ids)

    for _ in range(max_new_tokens):
        output = model(input_ids=input_ids, attention_mask=attention_mask)
        output_ids = torch.argmax(output.logits[:, -1], dim=-1)[None]
        input_ids = torch.cat([input_ids, output_ids], dim=1)
        attention_mask = torch.cat([attention_mask, torch.ones_like(output_ids)], dim=1)
    return input_ids

outputs = generate(**inputs, max_new_tokens=20)
tokenizer.decode(outputs, skip_special_tokens=True)[0]
```
