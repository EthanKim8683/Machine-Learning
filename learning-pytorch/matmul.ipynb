Strictly speaking, matmul only works on 2D matrices.

```python
import torch

a = torch.arange(1, 6 + 1).reshape(2, 3)
print("a =", a)

b = torch.arange(1, 12 + 1).reshape(3, 4)
print("b =", b)

print("a @ b =", a @ b)
```

Operands are unsqeezed into 2D matrices if necessary and the result is squeezed back. See examples below.

```python
a = torch.arange(1, 3 + 1)
print("a =", a)

b = torch.arange(1, 12 + 1).reshape(3, 4)
print("b =", b)

print("a @ b =", a @ b)
print("(a.unsqueeze(0) @ b).squeeze(0) =", (a.unsqueeze(0) @ b).squeeze(0))
```

```python
a = torch.arange(1, 6 + 1).reshape(2, 3)
print("a =", a)

b = torch.arange(1, 3 + 1)
print("b =", b)

print("a @ b =", a @ b)
print("(a @ b.unsqueeze(-1)).squeeze(-1) =", (a @ b.unsqueeze(-1)).squeeze(-1))
```

```python
a = torch.arange(1, 3 + 1)
print("a =", a)

b = torch.arange(1, 3 + 1)
print("b =", b)

print("a @ b =", a @ b)
print("(a.unsqueeze(0) @ b.unsqueeze(-1)).squeeze(0, -1) =", (a.unsqueeze(0) @ b.unsqueeze(-1)).squeeze(0, -1))
```

All but the last 2 dims are for batching 2D matmuls. Watch the shapes.

```python
a = torch.zeros((6, 7, 2, 3))
print("a.shape =", a.shape)

b = torch.zeros((6, 7, 3, 4))
print("b.shape =", b.shape)

print("(a @ b).shape =", (a @ b).shape)
```

If an operand is short on dims, it'll be broadcasted to match the batch dims of the other operand.

```python
a = torch.zeros((6, 9, 6, 9, 2, 3))
print("a.shape =", a.shape)

b = torch.zeros((6, 9, 3, 4))
print("b.shape =", b.shape)

print("(a @ b).shape =", (a @ b).shape)
print("(a @ b.broadcast_to(...)).shape =", (a @ b.broadcast_to(a.shape[:-2] + b.shape[-2:])).shape)
```
