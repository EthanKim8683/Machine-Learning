According to [PyTorch docs](https://docs.pytorch.org/docs/2.14/notes/broadcasting.html), "Two tensors are 'broadcastable' if the following rules hold: When iterating over the dimension sizes, starting at the trailing dimension, the dimension sizes must either be equal, one of them is 1, or one of them does not exist."

```python
import torch

x = torch.zeros((3, 1, 4))
x.shape
```

```python
x.broadcast_to(2, 4, 3, 1, 4).shape
```

```python
x.broadcast_to(3, 6, 4).shape
```

```python
# error since -1th dim is not 1
x.broadcast_to(3, 1, 8).shape
```
