```python
import torch

x = torch.zeros((3, 4), dtype=torch.long)
print("x =", x)

dim = 1
print("dim =", dim)

index = torch.randint(0, x.shape[dim], (*x.shape[:dim], 2, *x.shape[dim+1:]))
print("index =", index)

src = torch.arange(1, 12 + 1).reshape_as(x)
print("src =", src)

print("result =", x.scatter(dim, index, src))
```
