```python
from datasets import load_dataset

ds = load_dataset(path="data", data_files="swe_smith.parquet", split="train")
ds[0]
```

```python
from mlx_lm import load, generate
import json

from pathlib import Path
from hydra import compose, initialize_config_dir
from hydra.core.global_hydra import GlobalHydra

GlobalHydra.instance().clear()
with initialize_config_dir(version_base="1.2", config_dir=str(Path("conf").resolve())):
    cfg = compose(config_name="config", overrides=["env=local"])

model, tokenizer = load("mlx-community/Qwen2.5-Coder-7B-Instruct-4bit")
tokenizer.chat_template = cfg.tokenizer.chat_template

messages = json.loads(ds[0]["messages"])
prompt = tokenizer.apply_chat_template(
    messages[:-1] + messages[:2],
    add_generation_prompt=True,
    tokenize=False,
)
generate(model, tokenizer, prompt, max_tokens=1000, verbose=True)
```
