I got Cursor to write me a tracer. Let's try it out.

```python
from tracer import Tracer
import sys

t = Tracer()
with t.trace():
    x = 0

    def foo(n):
        global x
        x = 0
        for i in range(n):
            bar(i)
        print("foo")

    def bar(k):
        global x
        x = k
        print(k, file=sys.stderr)

    def baz(a, b):
        return a + b

    def qux(x):
        return 2 * x

    foo(5)
    print(baz(qux(1), qux(2)))

[trace] = t.traces()
print("*"*50, "Spans", "*"*50)
for span in trace.spans:
	print(span)
print("*"*50, "Stdout", "*"*50)
print(trace.stdout, end="")
print("*"*50, "Stderr", "*"*50)
print(trace.stderr, end="")
```
