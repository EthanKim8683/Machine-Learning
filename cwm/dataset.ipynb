```python
from tracer import Tracer

t = Tracer()
with t.trace():
	x = 0

	def foo(n):
		global x
		x = 0
		for i in range(n):
			bar(i)
		print("foo")

	def bar(k):
		global x
		x = k
		print(k)

	def baz(a, b):
		return a + b

	def qux(x):
		return 2 * x

	foo(5)
	print(baz(qux(1), qux(2)))

print(t.traces())
```
